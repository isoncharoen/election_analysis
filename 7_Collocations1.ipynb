```python
import os
import pandas as pd
import nltk
from nltk.collocations import BigramCollocationFinder
from nltk.metrics import BigramAssocMeasures
import matplotlib.pyplot as plt

file_path = '/Users/inqsoncharoen/Documents/Election_Analysis/combined_dems.csv'
data = pd.read_csv(file_path)

save_dir = "/Users/inqsoncharoen/Documents/Election_Analysis/Figures/CollocationPlots_Blue"
os.makedirs(save_dir, exist_ok=True) #make sure dir exists

def analyze_collocations_by_year(data, search_terms):
    results_by_year = {}
    for year, group in data.groupby('Year'):
        print(f"\nAnalyzing year: {year}")
        
        text_data = " ".join(group['Ads_Merged']).lower()
        tokens = nltk.word_tokenize(text_data)
        
        bigram_finder = BigramCollocationFinder.from_words(tokens)
        bigram_finder.apply_freq_filter(2)  
        collocations = bigram_finder.score_ngrams(BigramAssocMeasures.pmi)
        
        year_results = {}
        for search_term in search_terms:
            search_collocations = [(pair, score) for pair, score in collocations if search_term in pair]
            search_collocations = sorted(search_collocations, key=lambda x: x[1], reverse=True)[:7]
            year_results[search_term] = search_collocations
        
        results_by_year[year] = year_results
    
    return results_by_year

def visualize_collocations_by_year(results_by_year, search_terms):
    for year, terms in results_by_year.items():
        for search_term in search_terms:
            if search_term in terms and terms[search_term]:
                collocations = terms[search_term]
                words, scores = zip(*[(f"{pair[0]} {pair[1]}", score) for pair, score in collocations])
                
                # Plot
                plt.figure(figsize=(10, 6))
                plt.barh(words, scores, color='skyblue')
                plt.xlabel('PMI Score')
                plt.ylabel('Collocations')
                plt.title(f'Top collocations with "{search_term}" in {year}')
                plt.gca().invert_yaxis()
                
                filename = f"collocations_{search_term}_{year}.png".replace(" ", "_") #dynamic programming yay!!
                filepath = os.path.join(save_dir, filename)
                
                plt.savefig(filepath, dpi=300, bbox_inches="tight")
                plt.show()
                
                print(f"Figure saved: {filepath}")
            else:
                print(f"No significant collocations for '{search_term}' in {year}")

search_terms = ["american","border","country","crisis","cut","economy","freedom","health","immigrant","iraq","job","migrant","protect","right","tax","war","work"]
results_by_year = analyze_collocations_by_year(data, search_terms)

visualize_collocations_by_year(results_by_year, search_terms)
```

```python
import seaborn as sns

def heatmap_collocations_by_year(results_by_year, search_terms):
    #create df to store PMI scores across years
    heatmap_data = {}

    for year, terms in results_by_year.items():
        for search_term in search_terms:
            if search_term in terms and terms[search_term]:
                #get top collocation for the search term (highest pmi)
                best_collocation = terms[search_term][0]
                bigram, pmi_score = best_collocation
                heatmap_data.setdefault(search_term, {})[year] = pmi_score

    #convert to df
    heatmap_df = pd.DataFrame(heatmap_data).T
    heatmap_df = heatmap_df.fillna(0) #fill missing vals with 0

    plt.figure(figsize=(12, 8))
    sns.heatmap(heatmap_df, annot=True, cmap='coolwarm', cbar=True)
    plt.title('Heatmap of Bigram Collocations Across Years')
    plt.xlabel('Year')
    plt.ylabel('Search Terms')
    plt.show()
    
heatmap_collocations_by_year(results_by_year, search_terms)
```

```python
desired_ticks = [2000, 2004, 2008, 2012, 2016, 2020, 2024]

def temporal_line_plot_collocations(results_by_year, search_terms):
    plt.figure(figsize=(10, 6))

    for search_term in search_terms:
        years = []
        pmi_scores = []

        for year, terms in results_by_year.items():
            if search_term in terms and terms[search_term]: #top 7 bigrams for word?
                top_bigrams = terms[search_term]
                
                avg_pmi_score = sum(score for _, score in top_bigrams) / len(top_bigrams) #take avg pmi if more than one
                
                years.append(year)
                pmi_scores.append(avg_pmi_score)

        if years:
            plt.plot(years, pmi_scores, marker='o', label=search_term)

    plt.xlabel('Year')
    plt.ylabel('Average PMI score')
    plt.title('Average PMI score for top collocations')
    plt.xticks(desired_ticks)
    plt.legend(title="Keywords")
    plt.show()

temporal_line_plot_collocations(results_by_year, search_terms)
```

```python
def plot_collocation_heatmap(results_by_year, search_terms):
    for search_term in search_terms:
        print(f"\nProcessing search term: {search_term}")
        heatmap_data = {}
        
        #collect word pairs and their PMI scores by year
        for year, terms in results_by_year.items():
            if search_term in terms:
                for bigram, pmi_score in terms[search_term]:
                    word_pair = f"{bigram[0]} {bigram[1]}"
                    if word_pair not in heatmap_data:
                        heatmap_data[word_pair] = {}
                    heatmap_data[word_pair][year] = pmi_score
        
        print(f"Heatmap data for '{search_term}': {heatmap_data}")

        #skip this search term if there is no data
        if not heatmap_data:
            print(f"No data available for search term: {search_term}")
            continue

        #prepare data for the heatmap
        word_pairs = list(heatmap_data.keys())
        years = sorted(list(set(year for pair in word_pairs for year in heatmap_data[pair].keys())))

        if not years:
            print(f"No years with data for search term: {search_term}")
            continue

        matrix = np.zeros((len(word_pairs), len(years)))

        for i, word_pair in enumerate(word_pairs):
            for j, year in enumerate(years):
                matrix[i, j] = heatmap_data[word_pair].get(year, np.nan)
        
        if matrix.size == 0:
            print(f"Empty matrix for search term: {search_term}")
            continue

        df = pd.DataFrame(matrix, index=word_pairs, columns=years) #make df for heatmap?
        print(f"DataFrame for '{search_term}':\n{df}")
        plt.figure(figsize=(12, 8))
        sns.heatmap(df, cmap="YlGnBu", annot=True, fmt='.2f', cbar_kws={'label': 'PMI Score'}, vmin=0, vmax=np.nanmax(matrix))
        plt.title(f'Heatmap of Top Collocations for "{search_term}" Over Time')
        plt.xlabel('Year')
        plt.ylabel('Word Pairs (Collocations)')
        plt.xticks(rotation=45)
        plt.show()
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

def plot_word_clouds(results_by_year, search_terms, top_n=7): #loop through each search term and create a word cloud of top words
    for search_term in search_terms:
        plt.figure(figsize=(10, 8))
        all_words = {}

        #collect top bigrams for each year and search term
        for year, terms in results_by_year.items():
            if search_term in terms:
                top_bigrams = terms[search_term][:top_n]
                for bigram, pmi_score in top_bigrams:
                    word_pair = f"{bigram[0]} {bigram[1]}"
                    all_words[word_pair] = pmi_score  #use pmi
 
        if all_words:
            wordcloud = WordCloud(
                width=800,
                height=400,
                background_color="black",  
                colormap="Blues",       
                relative_scaling=0.5, 
                max_font_size=100 
            ).generate_from_frequencies(all_words)

            plt.imshow(wordcloud, interpolation="bilinear")
            plt.axis("off")
            plt.title(f'Word Cloud of Top Collocations for "{search_term}" Over Time', color="white")
            plt.show()
        else:
            print(f"No valid bigrams found for '{search_term}' to generate a word cloud.")

plot_word_clouds(results_by_year, search_terms)
```

```python
import matplotlib.pyplot as plt
import numpy as np

def plot_word_associations_over_time(results_by_year, search_terms, top_n=7):
    plt.figure(figsize=(10, 20))
    
    for search_term in search_terms:
        years = []
        word_labels = []
        pmi_scores = []
        
        for year, terms in results_by_year.items():
            if search_term in terms:
                #get top n bigrams for the search term in this year
                top_bigrams = terms[search_term][:top_n]
                
                for bigram, pmi_score in top_bigrams:
                    word_labels.append(f"{bigram[0]} {bigram[1]}")
                    years.append(year)
                    pmi_scores.append(pmi_score)

        if years:
            plt.scatter(years, pmi_scores, label=search_term, alpha=0.6, s=100)

            for i, word in enumerate(word_labels): #label points w/ words
                plt.annotate(word, (years[i], pmi_scores[i]), fontsize=9, alpha=0.6)

    plt.xlabel('Year')
    plt.ylabel('PMI score')
    plt.title('Word associations (collocation scores)')
    plt.legend(title="Keywords")
    plt.grid(True)
    plt.xticks(desired_ticks)
    plt.show()

plot_word_associations_over_time(results_by_year, search_terms)
```
